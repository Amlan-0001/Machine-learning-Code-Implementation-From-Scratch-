```python
import numpy as np 
import pandas as pd 
import sklearn 

df = pd.read_csv("perceptron_binary_classification.csv")
df.info()

x = df[["Hours_Studied", "Attendance"]].values
y = df["Pass"].values

from sklearn.model_selection import train_test_split 
x_train, x_test, y_train , y_test = train_test_split(
    x , y , test_size = 0.2, random_state = 42 
)

k = 5 
y_pred =[]

for test_point in x_test:
    distance = np.sqrt(np.sum((x_train - test_point)**2 , axis=1) )
    indices = np.argsort(distance)  ## sorts in incresaing order np.argsort  -> gives the indices of sorted elements 
    indices = indices[:k]   ### scling upto index k 
    labels = y_train[indices]
    count_0 = np.sum(labels == 0)
    count_1 = np.sum(labels ==1 )
    if count_0 > count_1:
        pred = 0 
    if count_0 < count_1:
        pred = 1 
    y_pred.append(pred)

from sklearn.metrics import accuracy_score , recall_score , roc_auc_score, confusion_matrix
a= accuracy_score(y_pred , y_test)
b = recall_score(y_pred, y_test)
print(b)

    
```

```python
### implement pnn  
import numpy as np 
import pandas as pd 
import sklearn 

df = pd.read_csv("PNN_dataset.csv")
df.head(5)

x = df[["Feature_1","Feature_2"]].values
y = df["Class"].values

## we will do label encoding of class A , B , C althroow it is not required here 
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(df["Class"])
print(y)

### splititng the dataset 
from sklearn.model_selection import train_test_split
x_train , x_test, y_train , y_test = train_test_split(
    x , y , random_state = 42 , test_size= 0.20
)

## model generatin 
sigma = 1.0 
y_pred =[]
for test_point in x_test:
    distance = np.sqrt(np.sum((x_train - test_point)**2,axis=1))
    g = np.exp(-(distance)**2/(2* sigma**2))
    score_A = np.sum(g[y_train==0])
    score_B = np.sum(g[y_train==1])
    score_C = np.sum(g[y_train==3])
    if score_A > score_B and score_A > score_C:
        predict = 0 
    elif score_B> score_A and score_B> score_C:
        predict = 1 
    else :
        predict = 2 
    y_pred.append(predict)
##evaluation 
# Evaluation
from sklearn.metrics import accuracy_score, confusion_matrix

acc = accuracy_score(y_test, y_pred)

print("Accuracy:", acc)
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
