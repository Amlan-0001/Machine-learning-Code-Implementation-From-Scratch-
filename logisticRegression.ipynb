```python
import numpy as np 
import pandas as pd 
import sklearn 

df = pd.read_csv("perceptron_binary_classification.csv")
df.info()

x = df[["Hours_Studied" , "Attendance"]].values
y = df["Pass"].values

#splitting 
from sklearn.model_selection import train_test_split
x_train , x_test, y_train,y_test = train_test_split(
    x , y , train_size = 0.20 , random_state = 42 
)

## code for model logistic regression 
weight = np.zeros(x_train.shape[1])
bias = 1 
learning_rate = 0.01 
epochs = 100 
for epoch in range(epochs):
    z = x_train@weight + bias 
    A = 1 / 1+ np.exp(-z)
    #calculate the gradiant 
    error = A - y_train
    dw = x_train.T @ error / len(x_train)
    db = np.mean(error)

    # update weight 
    weight_new = weight - learning_rate*dw
    bias_new = bias - learning_rate* db 
## prediction on test data 
Z = np.dot(x_test, weight) + bias
A = 1 / (1 + np.exp(-Z))

y_pred = (A >= 0.5).astype(int)

### evalutaion 
from sklearn.metrics import accuracy_score , confusion_matrix
aa = accuracy_score(y_test , y_pred)
confusion_matrix(y_test , y_pred)

print(aa)
```
