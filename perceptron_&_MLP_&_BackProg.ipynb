```python
import numpy as np
import pandas as pd

# Load dataset
data = pd.read_csv("perceptron_binary_classification.csv")

data.info()

# Separate input and target
target = data["Pass"].values
X = data[["Attendance", "Hours_Studied"]].values

# Split dataset
from sklearn.model_selection import train_test_split

X_train, X_test, target_train, target_test = train_test_split(
    X, target, test_size=0.20, random_state=42
)

# Initialize parameters
weight = np.zeros(X_train.shape[1])
bias = 1
epochs = 100
learning_rate = 0.1

# Training the Perceptron
for epoch in range(epochs):

    for i in range(len(X_train)):

        # Calculate weighted sum
        z = np.dot(X_train[i], weight) + bias

        # Activation function
        if z >= 0:
            y_pred = 1
        else:
            y_pred = 0

        # Calculate error
        error = target_train[i] - y_pred

        # Update weights
        weight = weight + learning_rate * error * X_train[i]

        # Update bias
        bias = bias + learning_rate * error

# Display trained parameters
print("\nFinal weights:", weight)
print("Final bias:", bias)

# Make predictions
y_pred = []

for i in range(len(X_test)):

    z = np.dot(X_test[i], weight) + bias

    if z >= 0:
        prediction = 1
    else:
        prediction = 0

    y_pred.append(prediction)

y_pred = np.array(y_pred)

# Evaluate model
from sklearn.metrics import accuracy_score, confusion_matrix

accuracy = accuracy_score(target_test, y_pred)

print("\nAccuracy:", accuracy)
print("Confusion Matrix:\n", confusion_matrix(target_test, y_pred))
```

```python
y = df["Pass"].values 
x = df[["Hours_Studied" , "Attendance"]].values
```

```python
from sklearn.model_selection import train_test_split 
x_train , x_test , y_train , y_test = train_test_split(
    x , y , test_size = 0.20 , random_state = 42
)
```

```python
## Forward Propagation Implementation

learning_rate = 0.1
epochs = 100

# Architecture: 2 Input Features -> 3 Hidden Neurons -> 1 Output Neuron

# Initialize hidden layer weights (2 inputs, 3 neurons)
weight_hidden = np.random.randn(x_train.shape[1], 3) * 0.01

# Hidden layer bias (1 row, 3 neurons)
bias_hidden = np.zeros((1, 3))

# Initialize output layer weights (3 hidden neurons, 1 output neuron)
weight_output = np.random.randn(3, 1) * 0.01

# Output layer bias (1 output neuron)
bias_output = np.zeros((1, 1))

# Store predictions for all training samples
y_pred = []

for epoch in range(epochs):

    for i in range(len(x_train)):

        # Hidden Layer
        z1 = np.dot(x_train[i], weight_hidden) + bias_hidden
        A1 = np.maximum(0, z1)  # ReLU activation

        # Output Layer
        z2 = np.dot(A1, weight_output) + bias_output
        A2 = 1 / (1 + np.exp(-z2))  # Sigmoid activation

        # Convert probability into class (0 or 1)
        if A2 >= 0.5:
            prediction = 1
        else:
            prediction = 0

        y_pred.append(prediction)

# Calculate Binary Cross-Entropy Loss
y_pred = np.array(y_pred)

print("Predictions:", y_pred)
```

```python

## MLP - Forward Propagation + Backpropagation

learning_rate = 0.1
epochs = 100

# Architecture: 2 Input Features -> 3 Hidden Neurons -> 1 Output Neuron

np.random.seed(42)

# Initialize weights and biases
weight_hidden = np.random.randn(x_train.shape[1], 3) * 0.01
bias_hidden = np.zeros((1, 3))

weight_output = np.random.randn(3, 1) * 0.01
bias_output = np.zeros((1, 1))


for epoch in range(epochs):

    # Store values of all training samples
    Z1_list = []
    A1_list = []
    A2_list = []

    ## FORWARD PROPAGATION

    for i in range(len(x_train)):

        # Hidden layer
        z1 = np.dot(x_train[i], weight_hidden) + bias_hidden
        A1 = np.maximum(0, z1)

        # Output layer
        z2 = np.dot(A1, weight_output) + bias_output
        A2 = 1 / (1 + np.exp(-z2))

        # Store intermediate values
        Z1_list.append(z1.flatten())
        A1_list.append(A1.flatten())
        A2_list.append(A2.item())

    # Convert stored values into NumPy arrays
    Z1 = np.array(Z1_list)
    A1 = np.array(A1_list)
    A2 = np.array(A2_list).reshape(-1, 1)

    Y = y_train.reshape(-1, 1)

    ## BACKPROPAGATION ********************************************************************************************

    # Output layer error
    dZ2 = A2 - Y

    # Output layer gradients
    dW2 = np.dot(A1.T, dZ2) / len(x_train)
    db2 = np.sum(dZ2, axis=0, keepdims=True) / len(x_train)

    # Hidden layer error
    dZ1 = np.dot(dZ2, weight_output.T) * (Z1 > 0)

    # Hidden layer gradients
    dW1 = np.dot(x_train.T, dZ1) / len(x_train)
    db1 = np.sum(dZ1, axis=0, keepdims=True) / len(x_train)

    ## UPDATE WEIGHTS AND BIASES

    weight_hidden = weight_hidden - learning_rate * dW1
    bias_hidden = bias_hidden - learning_rate * db1

    weight_output = weight_output - learning_rate * dW2
    bias_output = bias_output - learning_rate * db2


## FINAL PREDICTION ON TEST DATA

y_pred = []

for i in range(len(x_test)):

    z1 = np.dot(x_test[i], weight_hidden) + bias_hidden
    A1 = np.maximum(0, z1)

    z2 = np.dot(A1, weight_output) + bias_output
    A2 = 1 / (1 + np.exp(-z2))

    if A2 >= 0.5:
        y_pred.append(1)
    else:
        y_pred.append(0)


## MODEL EVALUATION

from sklearn.metrics import accuracy_score, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
