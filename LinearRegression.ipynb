```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df = pd.DataFrame({"X": [22, 25, 47, 52, 46, 56, 23, 40, 60, 36, 29, 49],
    "Y": [25000, 32000, 80000, 90000, 75000, 95000, 28000, 65000, 110000, 58000, 40000, 85000]})
```

```python
X = df[["X"]].values
y = df["Y"].values
```

```python
X = np.column_stack((np.ones(len(X)), X))
```

```python
X_transpose = X.T
A = X_transpose @X
B = X_transpose @ y
```

```python
# find the inverse 
A_inv = np.linalg.inv(A)
```

```python
theta = A_inv @B
```

```python
y_pred = X @ theta 
```

```python
from sklearn.metrics import(
    mean_squared_error , r2_score 
)
```

```python
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)
print(mse, r2)
```
