```python
### fuzzy c 


import numpy as np
import pandas as pd
import skfuzzy as fuzz

# Step 1: Load dataset
df = pd.read_csv("PCA_Dataset.csv")

# Step 2: Separate features
X = df.drop("target", axis=1).values

# Step 3: Apply Fuzzy C-Means
centers, membership, _, _, _, iterations, fpc = fuzz.cluster.cmeans(
    X.T, c=3, m=2, error=0.005, maxiter=1000
)

# Step 4: Assign clusters
clusters = np.argmax(membership, axis=0)

# Step 5: Display results
print("Centroids:\n", centers)
print("Membership Matrix:\n", membership)
print("Cluster Labels:\n", clusters)
print("Iterations:", iterations)


#pip install scikit-fuzzy
```

```python

import numpy as np
import pandas as pd

# Step 1: Load dataset
df = pd.read_csv("PCA_Dataset.csv")  # Read CSV file
X = df.drop("target", axis=1).values  # Remove target column

# Step 2: Initialize centroids
K = 3  # Number of clusters
centroids = X[:K].copy()  # Select first 3 rows as centroids

# Step 3: Repeat K-Means
for i in range(10):  # Repeat 10 times

    clusters = []  # Store cluster number of each point

    # Assign each point to nearest centroid
    for point in X:
        d1 = np.linalg.norm(point - centroids[0])  # Distance from centroid 1
        d2 = np.linalg.norm(point - centroids[1])  # Distance from centroid 2
        d3 = np.linalg.norm(point - centroids[2])  # Distance from centroid 3

        distances = [d1, d2, d3]  # Store all distances
        cluster = distances.index(min(distances))  # Find nearest centroid
        clusters.append(cluster)  # Assign cluster to point

    # Update centroids
    for j in range(K):
        points = X[np.array(clusters) == j]  # Select points of cluster j

        if len(points) > 0:  # Check if cluster is not empty
            centroids[j] = np.mean(points, axis=0)  # Calculate new centroid

# Step 4: Display results
print("Centroids:", centroids)  # Display final centroids
print("Clusters:", clusters)  # Display cluster assignments
```

```python
## Aglomorative 

import pandas as pd
from sklearn.cluster import AgglomerativeClustering

# 1. Load dataset
df = pd.read_csv("PCA_Dataset.csv")
X = df.drop("target", axis=1)

# 2. Create model
model = AgglomerativeClustering(n_clusters=3)

# 3. Fit and predict clusters
clusters = model.fit_predict(X)

# 4. Display results
print("Cluster Labels:", clusters)
```

```python
### Aglo

import numpy as np
import pandas as pd

# Step 1: Load dataset
df = pd.read_csv("PCA_Dataset.csv")
X = df.drop("target", axis=1).values

# Step 2: Make each data point a separate cluster
clusters = [[i] for i in range(len(X))]

# Step 3: Merge clusters until only 3 remain
while len(clusters) > 3:

    min_dist = float('inf')

    # Find nearest two clusters
    for i in range(len(clusters)):
        for j in range(i + 1, len(clusters)):

            # Calculate centre of both clusters
            c1 = np.mean(X[clusters[i]], axis=0)
            c2 = np.mean(X[clusters[j]], axis=0)

            # Calculate distance between centres
            d = np.sqrt(np.sum((c1 - c2) ** 2))

            if d < min_dist:
                min_dist = d
                a = i
                b = j

    # Merge the nearest clusters
    clusters[a] = clusters[a] + clusters[b]
    clusters.pop(b)

# Step 4: Display final clusters
print("Final Clusters:", clusters)
```
